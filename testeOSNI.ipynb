```python
ordens = dados["orders"]

print("Quantidade de ordens:", len(ordens))
```

```python
for ordem in ordens:
    print(
        f"ID: {ordem['_id']} | "
        f"Origem: {ordem['origin']['location']['name']} | "
        f"Destino: {ordem['destination']['location']['name']} | "
        f"Saída: {ordem['departure']} | "
        f"Chegada: {ordem['arrival']} | "
        f"Status: {ordem['status']}"
    )
```

```python
status_validos = {
    "requested",
    "in_transit",
    "ordered",
    "initiated"
}

ordens_validas = [
    ordem
    for ordem in ordens
    if ordem.get("status") in status_validos
]

print("Ordens retornadas:", len(ordens))
print("Ordens válidas:", len(ordens_validas))
```

```python
import pandas as pd

linhas = []

for ordem in ordens_validas:

    origem = ordem["origin"]["location"]["name"]
    destino = ordem["destination"]["location"]["name"]

    for item in ordem.get("items", []):

        linhas.append({
            "ordem": ordem["_id"],
            "origem": origem,
            "destino": destino,
            "status": ordem["status"],
            "departure": ordem["departure"],
            "arrival": ordem["arrival"],
            "sku": item["sku"],
            "nome": item["name"],
            "quantidade": item["quantity"]
        })

df = pd.DataFrame(linhas)

df.head()
```

```python
def classificar_fluxo(row):

    origem = row["origem"].lower()
    destino = row["destino"].lower()

    if "aldeia" in origem and "comfrio" in destino:
        return "Aldeia → Comfrio"

    if "aldeia" in origem:
        return "Aldeia → CD"

    if "comfrio" in origem:
        return "Comfrio → CD"

    return "Outros"


df["fluxo"] = df.apply(classificar_fluxo, axis=1)
```

```python
df["fluxo"].value_counts()
```

```python
aldeia_comfrio = df[
    df["fluxo"] == "Aldeia → Comfrio"
]

print(
    "Ordens:",
    aldeia_comfrio["ordem"].nunique()
)

print(
    "Unidades:",
    aldeia_comfrio["quantidade"].sum()
)
```

```python
aldeia_cd = df[
    df["fluxo"] == "Aldeia → CD"
]

print(
    "Ordens:",
    aldeia_cd ["ordem"].nunique()
)

print(
    "Unidades:",
    aldeia_cd ["quantidade"].sum()
)
```
